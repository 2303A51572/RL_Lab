```python

import gymnasium as gym
import numpy as np


def random_agent(environment_name, number_of_episodes=10):

    print("\n========================================")
    print("Environment:", environment_name)
    print("========================================")

    env = gym.make(environment_name)

    rewards = []

    for episode in range(number_of_episodes):

        # Reset the environment
        state, info = env.reset()

        total_reward = 0
        terminated = False
        truncated = False

        # Run the episode
        while not (terminated or truncated):

            # Select a random action
            action = env.action_space.sample()

            # Execute the action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Add reward
            total_reward += reward

            # Update state
            state = next_state

        rewards.append(total_reward)

        print(
            "Episode:", episode + 1,
            "Total Reward:", total_reward
        )

    env.close()

    # Calculate results
    average_reward = np.mean(rewards)

    print("\n----------- RESULTS -----------")
    print("Total Episodes :", number_of_episodes)
    print("Rewards        :", rewards)
    print("Average Reward :", average_reward)
    print("Maximum Reward :", max(rewards))
    print("Minimum Reward :", min(rewards))


# =====================================================
# CARTPOLE
# =====================================================

print("\n\n********** CARTPOLE **********")

random_agent("CartPole-v1", 10)


# =====================================================
# MOUNTAIN CAR
# =====================================================

print("\n\n********** MOUNTAIN CAR **********")

random_agent("MountainCar-v0", 10)


# =====================================================
# FROZEN LAKE
# =====================================================

print("\n\n********** FROZEN LAKE **********")

random_agent("FrozenLake-v1", 10)
```
