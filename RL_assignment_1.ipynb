```python
n = int(input("Enter number of machines: "))

q = {}

for i in range(n):
    name = input("Enter machine name: ")
    r = list(map(int, input("Enter rewards: ").split()))
    q[name] = sum(r) / len(r)

for name in q:
    print("Machine", name)
    print("Q(" + name + ") =", q[name])

best = max(q, key=q.get)
print("Greedy Selected Machine:", best)
```

```python
machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

q = {"A": 0, "B": 0, "C": 0}
n = {"A": 0, "B": 0, "C": 0}

for machine in machines:
    for r in machines[machine]:
        n[machine] += 1
        q[machine] = q[machine] + (r - q[machine]) / n[machine]

        print("Machine:", machine, "Reward:", r, "Q =", q[machine])

best = max(q, key=q.get)

print("\nFinal Q values:")
for machine in q:
    print("Q(" + machine + ") =", q[machine])

print("Greedy Selected Machine:", best)
```
